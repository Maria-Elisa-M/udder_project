```python
import os
import numpy as np
import pandas as pd
```

```python
dirpath = r"D:\udder_project_02\pred_labels"
fc_dir = os.path.join(dirpath, "frames_to_save")
sg_dir = os.path.join(dirpath, "segments")
kp_dir = os.path.join(dirpath, "keypoints")
ws_dir = os.path.join(dirpath, "watershed_segments")
ws_file = pd.read_csv(r"D:\udder_project_02\pred_labels\ws_class_predictions.csv")
feature_file = ""
```

```python
# step
print("Original")
# files
files = os.listdir(fc_dir)
cows =np.unique([file.split("_")[0] for file in files])
# cows
print(f"\tnumber of cows {len(files)}")
print(f"\tnumber of files {len(cows)}")
# lost cows
cow_set = set(cows)
lost_cows = []
for file in files:
    filepath = os.path.join(fc_dir, file)
    cow = file.split("_")[0]
    with open(filepath, "r") as f:
        imgs = len(f.read().split(","))
        if imgs <1:
            lost_cows.append(cow)
print(lost_cows)
# step
print("segmentation")
# files
files = os.listdir(sg_dir)
cows =np.unique([file.split("_")[0] for file in files])
# cows
print(f"\tnumber of cows {len(files)}")
print(f"\tnumber of files {len(cows)}")
# lost cows
step_cow_set = set(cows)
lost_cows = cow_set.difference(step_cow_set)
print(lost_cows)

# step
print("keypoints")
# files
files = os.listdir(kp_dir)
cows =np.unique([file.split("_")[0] for file in files])
# cows
print(f"\tnumber of cows {len(files)}")
print(f"\tnumber of files {len(cows)}")
# lost cows 
step_cow_set = set(cows)
lost_cows = cow_set.difference(step_cow_set)
print(lost_cows)

# qc
print("quality control")
good = ws_file[ws_file.thr09 ==1]
cows = np.unique(good.cow).astype("str")
# cows
print(f"\tnumber of files {len(good)}")
print(f"\tnumber of cows {len(cows)}")
step_cow_set = set(cows)
lost_cows = cow_set.difference(step_cow_set)
print(lost_cows)
```
