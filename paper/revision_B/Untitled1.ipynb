```python
import os
import pandas as pd
import numpy as np
from tifffile import imwrite
import json
```

```python
dirpath = r"D:\udder_project_02\features_dict"
teat_files = os.listdir(os.path.join(dirpath, "teat_length"))
shape_files = os.listdir(os.path.join(dirpath, "shape"))
keypoints_files = [file.replace("txt", "json") for file in os.listdir(r"D:\udder_project_02\pred_labels\keypoints")]
```

```python
teat_set = set(teat_files)
shape_set = set(shape_files)
kp_set = set(keypoints_files)
```

```python
missing_images = teat_set.difference(shape_set)
```

```python
len(teat_set.difference(kp_set))
```

```python
output_path = r"D:\udder_project_02"
array_path = os.path.join(output_path,"arrays")
depthpath = os.path.join(output_path, "depth_images")
```

```python
image_list = [file.replace(".json", "") for file in missing_images]
for image in image_list:
    frame = int(image.split("_")[-1])
    array_name = "_".join(image.split("_")[:-2]) + ".npy"
    frame_name= image + ".tif"
    src = os.path.join(array_path, array_name)
    depth_array = np.load(src, mmap_mode="r")
    if frame >= (len(depth_array)-1):
        print(f"1: {frame}")
        frame2 = frame - (frame - (len(depth_array)))-1
        print(f"2: {frame2}")
        imwrite(os.path.join(depthpath, frame_name), depth_array[frame2])
    else:
        imwrite(os.path.join(depthpath, frame_name), depth_array[frame])
```
