```python
from operator import mul

def time_tomin(my_time):
    factors = (60, 1, 1/60)
    t1 = sum(i*j for i, j in zip(map(int, my_time.split(':')), factors))
    t2 = sum(map(mul, map(int, my_time.split(':')), factors))
    return t2
```
