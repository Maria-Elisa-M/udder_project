```python
import os
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
plt.rcParams.update({'font.size': 14})
```

```python
dirpath = os.getcwd()
features_path = r"C:\Users\marie\rep_codes\udder_project\udder_processing\features_dict\gmfeature_table.csv"
data_path = r"C:\Users\marie\rep_codes\udder_project\delpro_vms\data"
plot_dir = os.path.join(os.path.normpath(dirpath + os.sep + os.pardir),r"adsa\examples")
```

```python
merged = pd.read_csv(os.path.join("long_format_df", "visit_ft_long_udder.csv"))
merged.columns
```

```python
u = "area_udder"
v = 'interval_sec'
selected = merged[[v, u ]].dropna(axis=0) 
res = stats.pearsonr(selected[v], selected[u])
plt.plot(selected[v]/3600, selected[u]/1000, '.', color = '#A32A44')
print(res)
plt.ylabel("Udder area (1000 pixels)")
plt.xlabel("Time since last visit (hours)")
plt.savefig(os.path.join(plot_dir, "time_area_lv.png"),bbox_inches='tight')
```
